```python
df = pd.DataFrame(articles)
df["date"] = pd.to_datetime(df["date"])
df.to_csv("articles.csv", index=False)
# to psql
```

```python
import pandas as pd
import numpy as np

# 假设你的 DataFrame 叫 df
table_name = 'articles_table' # 替换为你的真实表名
columns = ', '.join(df.columns)

# 生成 sql 语句列表
with open('insert_articles.sql', 'w', encoding='utf-8') as f:
    for index, row in df.iterrows():
        values = []
        for val in row:
            if pd.isna(val): # 处理空值 (NaN, None)
                values.append('NULL')
            elif isinstance(val, str):
                # SQL 中单引号需要转义为两个单引号
                val_escaped = val.replace("'", "''")
                values.append(f"'{val_escaped}'")
            # 针对时间类型的特殊处理（可选）
            elif isinstance(val, pd.Timestamp):
                values.append(f"'{val.strftime('%Y-%m-%d %H:%M:%S')}'")
            else:
                # 数字和布尔值等
                values.append(str(val))
        
        values_str = ', '.join(values)
        insert_sql = f"INSERT INTO {table_name} ({columns}) VALUES ({values_str});\n"
        f.write(insert_sql)

print("SQL文件已生成：insert_articles.sql")
```
